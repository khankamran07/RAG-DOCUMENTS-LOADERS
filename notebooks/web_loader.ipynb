### Recursive Loader

```python
from langchain_community.document_loaders import RecursiveUrlLoader
```

```python
# recursive loader

base_url = "https://docs.langchain.com/oss/python/integrations/document_loaders"

loader = RecursiveUrlLoader(url = base_url,
                             max_depth=2)
```

```python
# load the documents

documents = loader.load()
```

```python
len(documents)
```

```python
# metadata for 10 docs

for i in range(10):
    doc = documents[i]
    print(doc.metadata,end ="\n\n")
```

### lazy load

- Instead of returning a list of documents, it returns a generator object.
- You can loop through the generator to fetch the data efficiently.
- Generators are functions that use the yield statement instead of the return statement.
- This means you don't load all the data into memory at once.
- As you loop through the generator, it processes the data item by item.
- Each item is temporarily loaded into memory, processed, and then you move to the next item.

```python
documents_lazy_load = loader.lazy_load()
# Lazy loading will return an iterator or a generator to us.
```

```python
documents_lazy_load
```

```python
# Through lazy loading, I don't want to wait 3 minutes for all 80 documents to load at once.
```

```python
# Let's say I fetch the page content of the first element and load the first 500 characters.

documents[0].page_content[:500]
```

```python

counter = 0

for document in documents_lazy_load:
    # I want to add a stopping condition here. 
    if counter == 20:
        break
    # increment the counter
    counter +=1

    # print metadata and page content
    print(document.page_content[0:300])
    pp(document.metadata)

   # Here, I am viewing the page content and metadata of 20 documents.
```
